# Data-quality analysis
This notebook reads structured quality/audit artefacts and uses no validation business logic.

```python
import os
from pathlib import Path
import duckdb
root = Path(os.getenv('CLIMATE_ARTIFACT_ROOT', 'data'))
db_path = root / 'audit.duckdb'
if not db_path.exists(): raise FileNotFoundError(f'Run a pipeline first; missing {db_path}')
with duckdb.connect(str(db_path), read_only=True) as db:
    display(db.sql('SELECT severity, code, count(*) AS incidents FROM quality_incidents GROUP BY 1,2').df())
```
