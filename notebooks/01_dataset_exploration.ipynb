# Dataset exploration
Explore only saved canonical and curated artefacts; acquisition remains in the package CLI.

```python
import os, json
from pathlib import Path
import pandas as pd
from climate_pipeline.config import load_config
root = Path(os.getenv('CLIMATE_ARTIFACT_ROOT', 'data'))
config = load_config()
manifest = root / 'manifests' / f"{config['dataset_version']}.json"
if not manifest.exists(): raise FileNotFoundError(f'Run acquire-data first; missing {manifest}')
pd.DataFrame(json.loads(manifest.read_text())['entries'])[['location_id','year','sha256']].head()
```
